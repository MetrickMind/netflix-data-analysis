````python
import pandas as pd
df = pd.read_csv("netflix_titles.csv")
df.head()
````

````python
print(df.shape)
df.info()
````

````python
df["type"].value_counts()
````

````python
df["type"].value_counts().plot(kind="bar", title="Film və serial sayı")
````

````python
df["country"].value_counts().head(10)
````

````python
df["country"].value_counts().head(10).plot(kind="bar", title="Ən çox məhsul istehsal edən 10 ölkə")
````

````python
df["release_year"].value_counts().sort_index().plot(kind="line", title="İllərə görə say")
````

````python
# Hər sütunda neçə boş xana var
print(df.isnull().sum())

# Boş xanaları doldur
df["country"] = df["country"].fillna("Unknown")
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")

# Təkrarlanan sətirləri yoxla
print(df.duplicated().sum())
````

````python
genres = df["listed_in"].str.split(", ").explode()
genres.value_counts().head(10).plot(kind="barh", title="Ən çox 10 janr")
````
